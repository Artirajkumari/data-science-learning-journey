```python
x= df[['CLMSEX',	'CLMINSUR',	'SEATBELT',	'CLMAGE',	'LOSS']]
```

```python
y= df[['ATTORNEY']]
y
```

```python
#data validation 
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=12)
```

```python
x_train.shape, x_test.shape, y_train.shape,y_test.shape  # between  the x colmn column should be shame and between xx or y row shoud be shame
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
#Build the model
lr = LogisticRegression()
#Train the model
lr.fit(x_train,y_train)
```

```python
# Training Accuracy
y_train_prd= lr.predict(x_train)
```

```python
y_train_prd
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
confusion_matrix(y_train,y_train_prd)
```

```python
# acc = tn+tp/tn+fp+fn+tp = 0.71187
#precision class 0= tn/tn+fn=0.75 
# precesion class1= tp/fp+tp = 0.70
# recall/actual 0= tn/tnfp= 0.66
# recall class1= tp/fn+tp= 0.71
# f1_score= 2pr/p+r= 0.71
```

```python
print(classification_report(y_train,y_train_prd))
```

```python
y_test_prd= lr.predict(x_test)
```

```python
y_test_prd
```
