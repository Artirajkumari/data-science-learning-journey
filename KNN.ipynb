```python
x= df.iloc[:,:-1]

x
```

```python
y= df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size= 0.2,random_state= 28) 
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn= KNeighborsClassifier(n_neighbors=3)
knn. fit(x_train,y_train)

```

```python
#trainning accurcy
y_train_prd= knn.predict(x_train)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
confusion_matrix(y_train,y_train_prd)
```

```python
print(classification_report(y_train,y_train_prd))
```

```python
# test_accuracy
y_test_prd = knn.predict(x_test)
```

```python
confusion_matrix(y_test,y_test_prd)
```

```python
print(classification_report(y_test,y_test_prd))
```
