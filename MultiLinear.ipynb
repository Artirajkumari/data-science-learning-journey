```python
#Model Building - Separating x and y's #x1 = df.drop("MPG",axis=1,inplace=True) # row = 0, col = 1

x= df[['HP','VOL','SP','WT']]
y= df[['MPG']]
```

```python
x
```

```python
y
```

```python
#Data Validition : 80-20(0.2)
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=15)
#if we mention random state - it'll take same test data  and train data everytime. otherwise everytime train data and test data keeps changing.
```

```python
x_train, x_test, y_train,y_test
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
#x-train value should match with y-train value compulsary
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#Model Building
mlr = LinearRegression()
#Model Training
mlr.fit(x_train,y_train)
```

```python
#Testing Accuracy for Trained Data
y_train_pred = mlr.predict(x_train)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_train,y_train_pred) #always for know_data(train), we get highest prediction compared to unkowndata(test)
```

```python
#Testing Accuracy for Test Data
y_test_pred = mlr.predict(x_test)
```

```python
r2_score(y_test, y_test_pred)
```
